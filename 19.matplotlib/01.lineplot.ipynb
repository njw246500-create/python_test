```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
print(plt.style.available)
plt.style.use(['seaborn-v0_8'])
fig = plt.figure()
plt.axes()
```

- plt.plot(x축값, y축값)
- plt.plot(값) -> y축 값으로 인식, 0부터 y의 갯수만큼 1씩 증가 

```python
fig = plt.figure()
plt.plot([0,0.2,0.4,0.6,0.8,1]);
```

```python
x = np.arange(1,10,0.01)

fig = plt.figure()
plt.plot(x, np.sin(x))
```

```python
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
```

```python
plt.plot(np.random.randn(50).cumsum())
```
