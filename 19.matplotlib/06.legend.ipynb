```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use('default')
fig, ax = plt.subplots()
ax.plot(np.random.randn(10),'-r',label = 'A')
ax.plot(np.random.randn(10), '--g', label = 'B')
ax.plot(np.random.randn(10),':b', label = 'C')
ax.axis('equal')
ax.legend(loc = 'upper center', frameon=False, ncol=2)
```

```python
plt.figure(figsize = (8,4))
x = np.linspace(0,10,100)
y = np.cos(x[:, np.newaxis] * np.arange(0,2,0.2))
lines = plt.plot(x, y)
plt.legend(lines[:3], ['c1','c2','c3'])
```

```python
plt.plot(x,y[:,0], label = 'c1')
plt.plot(x,y[:,1], label = 'c2')
plt.plot(x,y[:,2], label = 'c3')
plt.plot(x,y[:,3:])

plt.legend(facecolor = 'lightgray', edgecolor = 'darkolivegreen', linewidth = 1)

```

```python
import matplotlib.pyplot as plt
import numpy as np

# Fixing random state for reproducibility
np.random.seed(19680801)

# Compute areas and colors
N = 150
r = 2 * np.random.rand(N)
theta = 2 * np.pi * np.random.rand(N)
area = 200 * r**2
colors = theta

fig = plt.figure()
ax = fig.add_subplot(projection='polar')
c = ax.scatter(theta, r, c=colors, s=area, cmap='hsv', alpha=0.75)
```

```python
x = np.linspace(1, 20, 100)
i = np.cos(x) - np.cos(x[:, np.newaxis])
plt.imshow(i, cmap = 'jet')
plt.colorbar(extend='both')
```

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

ax = plt.figure().add_subplot(projection='3d')
X, Y, Z = axes3d.get_test_data(0.05)

# Plot the 3D surface
ax.plot_surface(X, Y, Z, edgecolor='royalblue', lw=0.5, rstride=8, cstride=8,
                alpha=0.3)

# Plot projections of the contours for each dimension.  By choosing offsets
# that match the appropriate axes limits, the projected contours will sit on
# the 'walls' of the graph
ax.contourf(X, Y, Z, zdir='z', offset=-100, cmap='coolwarm')
ax.contourf(X, Y, Z, zdir='x', offset=-40, cmap='coolwarm')
ax.contourf(X, Y, Z, zdir='y', offset=40, cmap='coolwarm')

ax.set(xlim=(-40, 40), ylim=(-40, 40), zlim=(-100, 100),
       xlabel='X', ylabel='Y', zlabel='Z')

plt.show()
```
