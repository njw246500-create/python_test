﻿문제 1. 두 그래프를 나란히 그리기

주어진 데이터로 1행 2열 subplot 만들기
months = [1, 2, 3, 4, 5, 6]
sales_a = [20, 27, 25, 35, 42, 48]
sales_b = [17, 21, 30, 29, 37, 44]

* 요구사항
1. 왼쪽: A의 빨간색 실선, 원형 마커 / 오른쪽: B의 파란색 점선, 사각 마커.
2. 두 축 모두 xlim(1,6), ylim(0,55) 적용.
3. 각 그래프에 제목, x/y 레이블, legend() 추가.
4. 제목을 Product Sales Dashboard로 설정.

```python
months = [1, 2, 3, 4, 5, 6]
sales_a = [20, 27, 25, 35, 42, 48]
sales_b = [17, 21, 30, 29, 37, 44]
plt.rc('font', family = 'Malgun Gothic')
plt.subplot(1,2, 1)
plt.plot(months, sales_a, marker = 'o',color = 'red')
plt.title('a의 월 판매량')
plt.xlabel('월')
plt.ylabel('매출')
plt.xlim(1,6)
plt.ylim(0,55)
plt.subplot(1,2,2)
plt.plot(months,sales_b, marker = 's',color = 'blue')
plt.title('b의 월 판매량')
plt.xlabel('월')
plt.ylabel('매출')
plt.xlim(1,6)
plt.ylim(0,55)
```

문제 2. 아래 점수 데이터를 하나의 선 그래프로 그리고 실제 데이터 좌표를 이용해 최고점 표시
weeks = [1, 2, 3, 4, 5, 6]
scores = [65, 71, 78, 74, 89, 84]

* 요구사항
1. 보라색 선, 원형 마커, 제목에 Weekly Test Scores.
2. x축 범위 1~6, y축 범위 50~100.
3. 최고점 (5,89)에 annotate()으로 Peak: 89 표기.
4. 텍스트 위치는 데이터 좌표 (3.5, 96)으로 지정하고 화살표 모양은 arrowstyle='->'.

```python
weeks = [1, 2, 3, 4, 5, 6]
scores = [65, 71, 78, 74, 89, 84]

plt.title("Weekly Test Scores")
plt.xlim(1,6)
plt.ylim(50,100)
plt.plot(weeks, scores)
plt.annotate('Peak: 89',xy=(5,89), xytext = (3.5,96),arrowprops = dict(arrowstyle='->'))
```

문제 3. 두 지역의 온도 비교
days = [1, 2, 3, 4, 5, 6, 7]
city_a = [18, 20, 19, 23, 25, 24, 26]
city_b = [15, 17, 18, 20, 21, 23, 22]

* 요구사항
1. 두 곡선을 한 Axes에 그리고 스타일/범례를 다르게 설정.
2. 제목 City Temperatures, x축 Day, y축 Temperature (C).
3. annotate()를 사용하여 축 내부 왼쪽 위에 Weekly comparison 표시.
4. xy=(0.05, 0.9), xycoords=ax.transAxes를 사용해 축 기준 좌표로 배치.
5. 주석 화살표는 없이 작성

```python
days = [1, 2, 3, 4, 5, 6, 7]
city_a = [18, 20, 19, 23, 25, 24, 26]
city_b = [15, 17, 18, 20, 21, 23, 22]

fig, ax = plt.subplots()


plt.title("City Temperature")
plt.plot(days, city_a)
plt.plot(days, city_b)
plt.xlabel("Day")
plt.ylabel("Temperature (C)")
plt.annotate("Weekly comparison", xy = (2,24))
ax.text(0.05, 0.9,'transAxes(0.05, 0.9)', transform = ax.transAxes) # 요거 연습 다시
```


문제 4. 실적 데이터로 2행 1열 subplot을 그리고, 두 축이 아닌 그림 전체에 메모 배치
quarters = [1, 2, 3, 4]
revenue = [120, 145, 138, 175]
cost = [80, 86, 92, 96]

* 요구사항
1. 위쪽 축에 매출(revenue), 아래쪽 축에 비용(cost) 그래프.
2. 각 축에 제목/레이블/범례 추가, 선/마커를 다르게 지정.
3. fig.text()를 사용해 Figure 아래쪽 중앙에 Source: Sample business data를 표시.
4. 메모가 잘리지 않도록 여백 조정.

```python
quarters = [1, 2, 3, 4]
revenue = [120, 145, 138, 175]
cost = [80, 86, 92, 96]

fig, ax = plt.subplots()
fig.text(0.2, 0.2,'Source: Sample business data', transform=fig.transFigure)

plt.subplot(2,1,1)
plt.plot(quarters, revenue, 'r--o')
plt.title("revenue by quarters")
plt.xlabel("quarters")
plt.ylabel("revenue")

plt.subplot(2,1,2)
plt.plot(quarters, cost, 'b-.8')
plt.title("cost by quarters")
plt.xlabel("quarters")
plt.ylabel("cost")


```


문제 5. 다음 생산 데이터를 이용해 그래프 그리기 # 다시풀기!!
hours = [1, 2, 3, 4, 5, 6] 
line_a = [30, 42, 38, 55, 68, 72]
line_b = [25, 34, 45, 50, 57, 64]

* 요구사항
1. 첫 번째 축에는 A, B 두 생산 라인을 함께 그리기(범례, 마커, 선 스타일 사용).
2. 첫 축에서 A의 최고점 (6,72)을 xycoords=ax.transData로 지정하고 텍스트 (4,82)에서 arrowstyle='-|>' 화살표를 그리기.
3. 두 번째 축에는 line_a - line_b 차이 데이터를 선으로 그리기.
4. 두 번째 축 오른쪽 위에 (0.96, 0.92)라는 ax.transAxes 좌표로 Gap = A - B라는 안내문 표시(ha='right').
5. Figure 아래쪽 중앙 (0.5,0.03)에 fig.transFigure 좌표로 Production analysis | Sample data 문구 넣기.
6. 두 축 각각에 제목, x/y 레이블, 범위, 범례 지정. fig.suptitle() 및 글꼴 크기 설정.
7. 첫 번째 축에 데이터 (3,38)을 추가로 가리키는 arrowstyle='->' 주석을 작성해서 두 화살표 모양을 비교.

```python
hours = [1, 2, 3, 4, 5, 6]
line_a = [30, 42, 38, 55, 68, 72]
line_b = [25, 34, 45, 50, 57, 64]

# fig, ax = plt.subplots()

plt.subplot(2,1,1)
plt.plot(hours, line_a, 'r--o', label = 'A')
plt.xlabel('hours')
plt.ylabel('line')
plt.annotate('peak', xy = (6,72), xytext = (4,82), xycoords=ax.transData, arrowprops=dict(arrowstyle = '-|>'))
plt.plot(hours,line_b,'b-.8', label = 'B')
plt.xlabel('hours')
plt.ylabel('line')
plt.legend(loc = "upper left")

arr1 = np.array(line_a)
arr2 = np.array(line_b)
arr = arr1-arr2
plt.subplot(2,1,2)
plt.plot(hours, arr, 'r--o', label = 'A')
# plt.legend(bbox_to_anchor = (0, 1.0, 1., 0.1) , loc=0, ncol = 3, mode='extend', borderaxespad = 1.)
# plt.text(0.96,0.92,'Gap = A - B',ax.transAxes, ha = 'right')
```
