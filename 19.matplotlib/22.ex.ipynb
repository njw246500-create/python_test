﻿문제 1. 한 가게의 상품별 판매량 데이터를 사용해 막대그래프와 원그래프를 각각 그리기

sales = [120, 80, 60, 40]

* 요구사항
1. 그래프 2개를 나란히 배치
2. 첫 번째 그래프는 bar plot으로 상품별 판매량을 표시.
3. 막대 색상은 상품마다 다르게 지정하고, 제목과 축 이름을 추가

```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use(['seaborn-v0_8'])
```

```python
products = ['Coffee', 'Tea', 'Juice', 'Water']
sales = [120, 80, 60, 40]
y_pos = np.arange(len(products))
plo = [0.1 for _ in range(4)]
plt.subplot(1,2,1)
plt.bar(y_pos, sales, color = ['red', 'orange', 'blue', 'yellow'])
plt.title("sales by product")
plt.xlabel("products")
plt.ylabel("sales")

plt.subplot(1,2,2)
plt.pie(sales, explode = plo, startangle = 90, labels = plo, autopct = '%.1f%%',
    colors = ['red','blue','purple','gray','magenta']
    )
'''
height = [np.random.randn()*i for i in range(1,6)]
names = ['A', 'B', 'C', 'D', 'E']
y_pos = np.arange(len(names))
plt.bar(y_pos, height)
plt.xticks(y_pos, names, fontweight = 'bold')
plt.xlabel('group')'''

'''data = [10, 50, 30, 40, 60]
categories = ['a1','a2','a3','a4','a5']
plo = [0.1 for _ in range(5)] #_ : 사용하지 않는 변수
plt.pie(data, explode = plo, startangle = 90, labels = categories, autopct = '%.1f%%',
    colors = ['red','blue','purple','gray','magenta']
    )'''
```

2. 왼쪽에는 stem plot을 그리고 제목과 축 이름을 추가
3. 오른쪽에는 histogram을 그리고 구간 수를 5개로 지정
