```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = np.arange(-1, 1, 0.1)
x, y = np.meshgrid(a, a)
z = np.sin(x*y)

cs = plt.contour(x,y,z,levels = a)
plt.clabel(cs, inline = 2)
plt.colorbar();
```

```python
m = np.linspace(-1.0, 1.0, 1000)
x, y = np.meshgrid(m, m)
z = np.sqrt(x ** 2 + y ** 2 )
mv = np.linspace(z.reshape(-1,1).min(), z.reshape(-1,1).max(),40)

plt. contour(x, y, z, levels=mv)
plt.colorbar();
```

```python
plt.contourf(x,y,z,levels=mv)
plt.colorbar(); 
```

```python
plt.imshow(z, extent = [-1, 1, -1, 1], origin = 'lower',cmap = 'rainbow', alpha = 0.3)
```
