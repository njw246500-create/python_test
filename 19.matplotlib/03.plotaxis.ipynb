#### Plot Axis(플롯 축)

```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use(['seaborn-v0_8'])
```

```python
plt.plot(np.random.randn(50))
```

```python
plt.plot(np.random.randn(50))
plt.xlim(-1, 50) # x축의 범위 지정
plt.ylim(-4, 4)
```

```python
plt.plot(np.random.randn(50))
plt.axis([ -1 , 50, -5, 5]) # x,y의 범위를 한꺼번에 지정
```

```python
plt.plot(np.random.randn(50))
plt.axis('tight') # 위아래로 범위 줄여서 변동이 크게 보임
```

```python
plt.plot(np.random.randn(50))
plt.axis('equal') # 위아래 범위 늘려서 별로 변동이 없게 보임. 
```
