#### Error Bar 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,20,40)
dy = 1
y = np.sin(x) + dy * np.random.randn(40)

plt.errorbar(x,y, yerr=dy, fmt = 'H');
```

```python
plt.errorbar(x,y,yerr=dy,fmt='s',color = 'darkblue', ecolor = 'gray', elinewidth = 2, capsize = 2)
```
